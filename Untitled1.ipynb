```python

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 2. Create Patient Lifestyle Dataset
np.random.seed(42)

n = 300

df = pd.DataFrame({
    'Age': np.random.randint(20, 75, n),
    'BMI': np.random.normal(27, 5, n).round(1),
    'Sleep_Hours': np.random.normal(6.5, 1.5, n).round(1),
    'Exercise_Minutes': np.random.randint(0, 90, n),
    'Smoking': np.random.randint(0, 2, n),
    'Alcohol_Consumption': np.random.randint(0, 2, n),
    'Stress_Level': np.random.randint(1, 11, n),
    'Daily_Steps': np.random.randint(1000, 12000, n)
})

# 3. Display Dataset
print("PATIENT LIFESTYLE DATASET")
print(df.head(10))

print("\nDataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

# 4. Data Preprocessing
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# 5. Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(scaled_data)

# 6. Display Clustered Patients
print("\nCLUSTERED PATIENT DATA")
print(df.head(15))

print("\nNumber of Patients in Each Cluster:")
print(df['Cluster'].value_counts().sort_index())

# 7. Cluster-wise Lifestyle Analysis
print("\nCLUSTER-WISE AVERAGE LIFESTYLE")
print(df.groupby('Cluster').mean().round(2))

# 8. Silhouette Score
score = silhouette_score(scaled_data, df['Cluster'])
print("\nSilhouette Score:", round(score, 3))

# 9. Graph 1: Number of Patients in Each Cluster
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Cluster')
plt.title('Number of Patients in Each Lifestyle Cluster')
plt.xlabel('Cluster Group')
plt.ylabel('Number of Patients')
plt.show()

# 10. Graph 2: BMI vs Exercise
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='BMI',
    y='Exercise_Minutes',
    hue='Cluster',
    palette='Set1',
    s=70
)
plt.title('Patient Clustering: BMI vs Exercise')
plt.xlabel('Body Mass Index (BMI)')
plt.ylabel('Daily Exercise (Minutes)')
plt.show()

# 11. Graph 3: Lifestyle Heatmap
cluster_means = df.groupby('Cluster').mean()

plt.figure(figsize=(10, 6))
sns.heatmap(
    cluster_means,
    annot=True,
    cmap='coolwarm',
    fmt='.1f'
)
plt.title('Average Lifestyle Characteristics by Cluster')
plt.show()

# 12. Graph 4: PCA Cluster Visualization
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=pca_data[:, 0],
    y=pca_data[:, 1],
    hue=df['Cluster'],
    palette='Set1',
    s=70
)
plt.title('Patient Lifestyle Groups using K-Means and PCA')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

# 13. Final Result
print("\nFINAL RESULT")
print("Total Patients:", len(df))
print("Number of Clusters:", df['Cluster'].nunique())
print("Patient lifestyle clustering completed successfully!")
```
